```python
from pyspark.sql.types import(
    StructType,
    StructField,
    StringType,
    IntegerType
)

schema = StructType([
    StructField("id",IntegerType(),True),
    StructField("name",StringType(),True),
    StructField("age",IntegerType(),True),
    StructField("department",StringType(),True)
])

df = spark.createDataFrame(data, schema)
df.printSchema()
```

```python
df = spark.read.csv("/Volumes/my_files/my_schema/csv_files/airbnb.csv",header=True,inferSchema=True)
df.show(5, truncate=False)
```

```python
from pyspark.sql import functions as F

df = df.withColumn(
    "rating_numeric",F.when(F.col("rating") == "New", None).otherwise(F.col("rating").cast("double"))
)
```

```python
df = df.withColumn(
    "review_numeric",F.col("reviews").cast("int")
)
```

```python
df.select(
    "rating",
    "rating_numeric",
    "reviews",
    "review_numeric"
)
```

```python
spark.conf.set("spark.sql.ansi.enabled","false")
```

```python
# Find Highest Rated Properties

df.withColumn(
    "rating_numeric",
    F.when(F.col("rating") == "New", None).otherwise(F.expr("try_cast(rating as double)"))
).withColumn(
    "review_numeric",
    F.expr("try_cast(reviews as int)")
).filter(
    F.col("rating_numeric").isNotNull()
).select(
    "name",
    "country",
    "rating_numeric",
    "review_numeric",
    "price"
).orderBy(
    F.col("rating_numeric").desc()
).show(10)
```

```python
from pyspark.sql import functions as F

df = df.withColumn(
    "rating",
    F.expr("try_cast(rating AS DOUBLE)")
)
```

```python

df.filter(F.col("price") < 5000).show()
```

```python
df.filter((F.col("price") < 5000) & (F.col("rating") >= 4.5)).show()
```

```python
df.filter((F.col("price") < 5000) | (F.col("rating") >= 4.5)).show()
```

```python
df.filter(F.col("country").isin(["Turkey","Georgia","Norway"])).show()
```

```python
df.orderBy(F.col("price").asc()).show()
```

```python
df.orderBy(F.col("price").desc()).show()
```

```python
df.orderBy(
    F.col("price").asc(),
    F.col("rating").desc()
).show()
```

```python
df.agg(F.count("*").alias("total_listings")).show()
```

```python
df.agg(
    F.avg("price").alias("avg_price"),
    F.min("price").alias("min_price"),
    F.max("price").alias("max_price"),
).show()
```

```python
df.groupBy("country").count().show()
```

```python
df.groupBy("country").agg(
    F.count("*").alias("total_listings"),
    F.avg("price").alias("avg_price")).filter(
        F.col("total_listings") > 50
    ).orderBy(
        F.col("avg_price").desc()
    ).show(40)
```
